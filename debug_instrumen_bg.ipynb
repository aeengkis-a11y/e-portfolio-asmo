# Debug CSS Background Image
This notebook demonstrates and debugs why a CSS `background-image` may not display.

```python
from IPython.display import HTML, display
from pathlib import Path

# Section 1: Import Required Libraries
print('Imports ready')
```

## Define HTML with Background Image
Create an HTML snippet that uses CSS `background-image` and set the image URL or relative path.

```python
image_url = 'https://lh3.googleusercontent.com/d/1hOoFEulEiHcxVE_jz2WXqjY_rSefidTh=w1920'
html = f"""
<div style='width:100%;min-height:400px;background-image:url({image_url});background-size:cover;background-position:center;background-repeat:no-repeat;border:1px solid #444;'>
  <h2 style='color:#fff;padding:24px;text-shadow:0 0 12px rgba(0,0,0,.8)'>Background image test</h2>
</div>
"""
display(HTML(html))
```

## Display HTML in Notebook
Render the HTML snippet in the notebook to verify whether the background image appears.

```python
local_image = Path('instrumen-bg.jpg')
print('Local image exists:', local_image.exists())
print('Local image absolute path:', local_image.resolve() if local_image.exists() else 'N/A')

html_local = f"""
<div style='width:100%;min-height:400px;background-image:url({local_image.as_posix()});background-size:cover;background-position:center;background-repeat:no-repeat;border:1px solid #444;'>
  <h2 style='color:#fff;padding:24px;text-shadow:0 0 12px rgba(0,0,0,.8)'>Local image test</h2>
</div>
"""

display(HTML(html_local))
```

## Verify Image File Path
Use Python to confirm the image file exists and that the relative or absolute path used in CSS is correct.

```python
if local_image.exists():
    print('File size:', local_image.stat().st_size, 'bytes')
    print('File suffix:', local_image.suffix)
else:
    print('File is missing. Please download it to the notebook folder.')
```
